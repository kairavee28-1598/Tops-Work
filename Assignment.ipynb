```python
# Assignment 13
def reverse_string(s):
    
    # Base case
    if len(s) <= 1:
        return s

    # Recursive case
    return reverse_string(s[1:]) + s[0]

result = reverse_string("hello")

print("Original String:", "hello")
print("Reversed String:", result)


def sum_playlist_durations(durations):
    
    # Base case
    if len(durations) == 0:
        return 0

    # Recursive case
    return durations[0] + sum_playlist_durations(durations[1:])

durations = [180, 240, 210, 195, 300]

total_duration = sum_playlist_durations(durations)

print("Song Durations:", durations)
print("Total Duration:", total_duration, "seconds")
print("Total Duration:", total_duration // 60, "minutes",
      total_duration % 60, "seconds")


count = 10

def update_count():
    count = 5
    print("Inside:", count)

update_count()

print("Outside:", count)


def count_likes(posts):
    
    total = 0

    for post in posts.values():

        # Add likes of the current post
        total += post["likes"]

        # Add likes of all replies recursively
        total += count_likes(post.get("replies", {}))

    return total

instagram_posts = {
    "post1": {
        "likes": 100,
        "replies": {
            "reply1": {
                "likes": 20,
                "replies": {
                    "reply1_1": {
                        "likes": 10,
                        "replies": {}
                    }
                }
            },
            "reply2": {
                "likes": 15,
                "replies": {}
            }
        }
    },

    "post2": {
        "likes": 200,
        "replies": {
            "reply1": {
                "likes": 30,
                "replies": {}
            }
        }
    }
}


total_likes = count_likes(instagram_posts)

print("Total Instagram Likes:", total_likes)

# Global variable
app_status = "Offline"

def update_status():

    # Local variable
    user_status = "Online"

    print("Inside Function:")
    print("User Status:", user_status)
    print("App Status:", app_status)


print("Before Function Call:")
print("App Status:", app_status)

update_status()

print("After Function Call:")
print("App Status:", app_status)

# Trying to access local variable outside the function
# print(user_status)  # This would cause NameError
```

```python
# Assignment 20
class Product:
    def __init__(self, price):
        self._price = price

    def display_price(self):
        print("Product Price:", self._price)

# Create object
product1 = Product(1500)
product1.display_price()


class Product:
    def __init__(self, price):
        self._price = price

    # Getter method
    def get_price(self):
        return self._price

    # Setter method
    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative")
        self._price = new_price
    def display_price(self):
        print("Product Price:", self._price)

# Create object
product1 = Product(1500)

# Access price using getter
print("Current Price:", product1.get_price())

# Update price using setter
product1.set_price(2000)
print("Updated Price:", product1.get_price())

# Try assigning a negative price
try:
    product1.set_price(-500)
except ValueError as e:
    print("Error:", e)
    
    
class Playlist:
    def __init__(self):
        self._songs = []

    # Add a song
    def add_song(self, song):
        self._songs.append(song)
        print(song, "added to playlist")

    # Remove a song
    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)
            print(song, "removed from playlist")
        else:
            print(song, "not found in playlist")

    # Get current song list
    def get_songs(self):
        return self._songs.copy()

# Create playlist object
my_playlist = Playlist()

# Add songs
my_playlist.add_song("Kesariya")
my_playlist.add_song("Tum Hi Ho")
my_playlist.add_song("Apna Bana Le")

# Remove a song
my_playlist.remove_song("Tum Hi Ho")

# Display playlist
print("Current Playlist:", my_playlist.get_songs())


from abc import ABC, abstractmethod

# Abstract class
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

# Child class 1
class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"Processing ₹{amount} payment through UPI")

# Child class 2
class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Processing ₹{amount} payment through Credit Card")

# Create objects
upi_payment = UPI()
card_payment = CreditCard()

# Process payments
upi_payment.pay(1500)
card_payment.pay(2500)
```

```python
# Assignment 21
class Product:
    def get_discount(self):
        return 0

class Electronics(Product):
    def get_discount(self):
        return 10

# Create objects
product1 = Product()
product2 = Electronics()

print("Product Discount:", product1.get_discount(), "%")
print("Electronics Discount:", product2.get_discount(), "%")


class FoodOrder:
    def __init__(self, base_price):
        self.base_price = base_price
    def calculate_total(self):
        return self.base_price

class ZomatoOrder(FoodOrder):
    def calculate_total(self):
        delivery_charge = self.base_price * 0.05
        return self.base_price + delivery_charge

# Create objects
order1 = FoodOrder(1000)
order2 = ZomatoOrder(1000)

print("Food Order Total: ₹", order1.calculate_total())
print("Zomato Order Total: ₹", order2.calculate_total())


class Influencer:
    def bonus(self):
        return 2000

class BrandManager:
    def bonus(self):
        return 5000

# Common function
def show_bonus(employee):
    print("Employee Bonus: ₹", employee.bonus())

# Create objects
influencer1 = Influencer()
manager1 = BrandManager()

# Call the same function with different objects
show_bonus(influencer1)
show_bonus(manager1)


class User:
    def get_status(self):
        return "active"

class PremiumUser(User):
    def get_status(self):
        return "premium"

# Create objects
user1 = User()
user2 = PremiumUser()

# Print statuses
print("Regular User Status:", user1.get_status())
print("Premium User Status:", user2.get_status())
```
